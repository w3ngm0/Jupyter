### The loc() function 
- a label based data selecting method
- have to pass the name of the row or column which we want to select
- includes the last element of the range passed in it, unlike iloc().
- can accept the boolean data unlike iloc()



```python
df_cars = df_cars.T

display(df_cars.loc[2] )
display(df_cars.loc[5, 'Year'] )
display(df_cars.loc[6, ['Year', 'City']] )
```

### Operations that can be performed using the loc() method:
1. Selecting Data According to Some Conditions
2. Selecting a Range of Rows From the DataFrame
3. Updating the Value of Any Column

```python
# 1. selecting cars with brand 'Maruti' and Mileage > 25 
display(df_cars.loc[(df_cars.Make == 'Maruti') & (df_cars.Mileage > 25)]) 

# 2. selecting range of rows from 2 to 5 
display(df_cars.loc[2: 5]) 

# 3. updating values of Mileage if Year < 2015 
df_cars.loc[(df_cars.Year < 2015), ['Mileage']] = 22
display(df_cars) 
```

### The iloc function
- an indexed-based selecting method 
- have to pass an integer index in the method to select a specific row/column
- does not include the last element of the range passed in it unlike loc()
- iloc() does not accept the boolean data unlike loc()



```python
display( df_cars.iloc[3]) # row only specified
display( df_cars.iloc[3,2]) # row and column specified
```

### Operations that can performed using iloc():

1. Selecting Rows Using Integer Indices
2. Selecting a Range of Columns and Rows Simultaneously

```python
# 1. selecting 0th, 2nd, 4th, and 7th index rows 
display(df_cars.iloc[[0, 2, 4, 7]]) 

# 2. selecting rows from 1 to 4 and columns from 2 to 4 
display(df_cars.iloc[1: 5, 2: 5]) 
```
