## Statistical data normalization

Statistical data normalisation is the process of transforming numeric values to a common scale without distorting differences in the ranges of values.It helps ensure that features contribute equally to analysis and modelling, especially when they originally have very different units or scales. Normalisation is vital especially in machine learning algorithms that depend on distance, gradient or scale. e.g., clustering, neural networks

Consider the following features in a dataset about people.

- Age: 0–100
- Income: 0–200,000
- Distance: 0–500

If we use these directly in many algorithms, the feature with the largest scale dominates the calculations. Normalisation fixes this by putting everything on a comparable scale.

### Common normalisation techniques

- min-max scaling

  $ x_{norm}=\frac{x-x_{min}}{x_{max}-x_{min}}$
   
- z-score (standard) scaling

  $ z=\frac{x-\mu}{\sigma}$

- decimal scaling

  $ x_{norm}=\frac{x}{10^j}$

- vector normalisation

  $ x_{norm}=\frac{x}{||x||}$

In our context, data normalization refers to the process of transforming data into a common format and scale without distorting the differences in the ranges of values. This process is important because it ensures that data is consistent and can be easily compared and analysed.




### Normalization tasks

- Handling missing values
- Consistency: ensuring that data is in a consistent format (ranging from all dates being in the same format to ensuring that all names are in the same case)
- Data scaling: ensuring that data is on the same scale (e.g. converting all temperatures to Celsius)
- Encoding categorical data: converting categorical data into a numerical format and ensuring that the data is consistent

### Yeah, NaN - Handling missing values

In the real world, data is often incomplete. This can be due to various reasons, including human error, technical issues, or the nature of the data itself. Missing values can **interfere in the analysis and interpretation of data**. However, it is important to **handle missing values in a way that does not distort the data or introduce bias**.

#### Representing missing data

- Pandas represents missing data differently depending on type:
    - Numbers: `np.nan`
    - Boolean: `pd.NA`
    - Objects: `None`
    - Date: `pd.NaT`
- All are treated as `null`
- Unlike *bespoke* 'nulls', such as:
    - The string `"N/A"`
    - Empty strings `''` `" "`
    - 0
    - etc... 

#### Various forms of missing data

When we talk about missing data, we need to distinguish between data that Pandas has interpreted as missing and data that is missing in a more general sense. Empty strings and bespoke designations of missing data (e.g. `"N/A"`) are all examples of missing data that Pandas will not interpret as missing. We'll deal with them when we look at data cleaning more generally.

For now, we will deal with values that Pandas interprets as missing, or `null`. These can be represented by `NaN` (Not a Number) in Pandas when, well, the data type of the column is a number. `NaN` is a special value that is part of the IEEE floating-point specification. It is used to represent missing or undefined numerical data. The `numpy` library, which Pandas is built on, provides the `np.nan` representation, which is the value that Pandas uses to represent missing data.

Other data types may have their own representation of missing data within Pandas. For example, `NaT` is used to represent missing values in datetime columns, `pd.NA` is used to represent missing values in string columns.

#### Dealing with nulls

Pandas provides a number of methods for handling missing values, including:

- `isnull()`: returns a boolean mask indicating missing values
- `notnull()`: returns the opposite of `isnull()`
- `dropna()`: returns a filtered version of the data, excluding missing values
- `fillna()`: returns a copy of the data with missing values filled or imputed\*

\*imputing refers to the process of replacing missing values with a statistical estimate of the missing value

### Converting data types

Pandas provides a number of methods for converting data types, including:

- `astype()`: converts the data type of a column
- `to_datetime()`: converts a column to a datetime data type
- `to_numeric()`: converts a column to a numeric data type

Let's look at the following example data set and try and convert its types as well as deal with any "dirtyness" in the data that might hamper conversion

````python
data = [['Sydney', '2023-01', 120.5, 25.3, 5.2],
  ['Melbourne', '2023-01', '', 20.1, 7.8],
  ['Brisbane', '2023-01', 200.2, 27.5, None],
  ['Perth', '2023-01', 0.0, 30.4, 8.9],
  ['Adelaide', '2023-01', 15.2, 23.0, 7.5],
  ['Hobart', '2023-01', 47.5, 15.6, 4.3],
  ['Darwin', '2023-01', 250.3, 29.0, 6.1],
  ['Canberra', 'January 2023', 60.4, 20.0, 4.8],
  ['Sydney', '2023-02', 138.2, 25.8, 6.2],
  ['Melbourne_AU', '2023-02', np.nan, 22.3, 7.1],
  ['Brisbane_AU', '2023-02', 220.4, 28.0, None],
  ['Perth_AU', '2023-02', "NA", 32.2, 9.0],
  ['Adelaide_AU', '2023-02', 12.8, 24.5, 7.0],
  ['Hobart_AU', '2023-02', 52.7, 17.2, 5.1],
  ['Darwin_AU', '2023-02', 300.5, 28.5, 6.5],
  ['Canberra_AU', 'February 2023', 70.6, 21.5, 5.4]]

columns = ['City',
  'Month',
  'Rainfall (mm)',
  'Average Temperature (C)',
  'Average Wind Speed (km/h)']

df = pd.DataFrame(data=data, columns=columns)
df.head()
````

Let's begin by trying to convert the `Month` column to a datetime data type. We can use the `to_datetime()` method to do this: 

````python
pd.to_datetime(df['Month'])
````

We got an error! Don't get discouraged - remember to look at the very bottom of the message to see what the error is. 

In this case, it is a `ValueError` and it says that the month is not in the correct format.

We can use the `errors` parameter to handle this. The `errors` parameter can take the following values:

- `raise`: this is the default value and will raise an error if the conversion fails. Just like it did now
- `coerce`: this will convert the column to datetime and set any errors to `NaT`
- `ignore`: this will return the original column if the conversion fails

None are that great in this case. If we use `ignore` we end up with don't end up with a date. If we use `coerce` we end up with a date but we lose the month.

The following code provides a simplified example where the conversion is successful (can anyone figure out why it works in this case??). 

````python

data_simple_adjustment = [['Sydney', '2023-01', 120.5, 25.3, 5.2],
                          ['Melbourne', '2023-01', '', 20.1, 7.8],
                          ['Brisbane', '2023-01', 200.2, 27.5, None],
                          ['Perth', '2023-01', 0.0, 30.4, 8.9],
                          ['Adelaide', '2023-01', 15.2, 23.0, 7.5],
                          ['Hobart', '2023-01', 47.5, 15.6, 4.3],
                          ['Darwin', '2023-01', 250.3, 29.0, 6.1],
                          ['Canberra', 'February 2023', 60.4, 20.0, 4.8]]  # This entry is different


df_simple_adjustment = pd.DataFrame(data=data_simple_adjustment, columns=columns)


df_simple_adjustment['Month'] = pd.to_datetime(df_simple_adjustment['Month'], errors='coerce')

df_simple_adjustment.dtypes
````

Any **consistent** date format can be handled by `pd.to_datetime`: the correct format is either inferred or can be entered manually using format specifiers.

When there are *inconsistent* date formats, we may have to resort to a more traditional Python approach.

For example, here, I used a function to identify when the date format contains the Month and manually fixes it:

````python
def standardize_month_format(month):
 
    if "-" in month:
        return month
 
    month_map = {"January": "01", "February": "02", "March": "03", "April": "04", "May": "05",
                 "June": "06", "July": "07", "August": "08", "September": "09", "October": "10",
                 "November": "11", "December": "12"}
    parts = month.split() # e.g. split January 2023 to ['January', '2023']
    month_num = month_map[parts[0]]
    year = parts[1]
    return f"{year}-{month_num}"
````

We can then use this function to clean up the inconsistencies: 


````python
df['Month'] = df['Month'].apply(standardize_month_format)

df['Month'] = pd.to_datetime(df['Month'], format='%Y-%m') # format specifier is optional demo only

df.head()
````

````python
df.dtypes
````

#### Numeric fields

Why is rainfall an object and not numeric?

This dataset is small enough that you probably know. But what if it wasn't?

On one level, we can just `coerce` the data into a numeric field. But what if there was information in those fields (maybe it says `9mm`?)


````python
pd.to_numeric(df['Rainfall (mm)'], errors='coerce')
````

#### Optional: Advanced
The following code uses isna as a mask to find those values that would be coerced to `NaN`. In this case, we can safely convert them.

````python
df['Rainfall (mm)'][pd.to_numeric(df['Rainfall (mm)'], errors='coerce').isna()] 
````

#### Filling missing data

After speaking to the subject matter expert (SME), we decide to make all the NaNs for rainfall zero.

````python
# 1. Convert rainfall to numeric:

df['Rainfall (mm)'] = pd.to_numeric(df['Rainfall (mm)'], errors='coerce')

# 2. Convert the NaNs to 0

df['Rainfall (mm)'].fillna(0)
````

Notice that the change is ephemeral. We can make it permanent by either assigning the series to the rainfall column or by using `inplace=True`:

````python
df['Rainfall (mm)'].fillna(0, inplace=True)
````

## Visualisation

The module `matplotlib` is a popular data visualisation library. It is often used  with `pandas` to create plots. A more modern library is `seaborn` which is built on top of `matplotlib` and provides a high-level interface for drawing attractive and informative statistical graphics.

Let's look at our simple data set and see if we can create some visualisations.

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
sns.histplot(data=df, x='Average Temperature (C)', kde=True)
plt.show()
````

````python
sns.barplot(data=df, x='City', y='Rainfall (mm)')
plt.show()
````

````python
sns.scatterplot(data=df, x='Average Temperature (C)', y='Average Wind Speed (km/h)')
plt.show()
````

````python
sns.set_theme(style="darkgrid")
sns.scatterplot(data=df, x='Average Temperature (C)', y='Average Wind Speed (km/h)')
plt.title('Temperature vs. Wind Speed')
plt.xlabel('Average Temperature (°C)')
plt.ylabel('Average Wind Speed (km/h)')
plt.show()
````
