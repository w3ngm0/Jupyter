### Creating a DataFrame

You can create data frames from various sources, but recall that, unlike a Series, we need to specify **two dimensions**, which can be represented by a collection of series.

A common pattern is to create a DataFrame from a dictionary where the keys represent the column names and the values represent a series. 

The following figure highlights this relationship:

![by default, a dictionary key maps to a column](images/key_is_col.png)

Have a go at defining your first DataFrame this way:

````python
cities = {
    'population': [
        2.04, 
        5.23, 
        4.96, 
        2.19, 
        1.29
    ], 
    'area': [
        5386, 
        2058, 
        9993, 
        5905, 
        1826
    ]
}

df = pd.DataFrame(cities)

df
````

````python
# notice that the key is transformed to the column, a type we did not have when we learned about a series.
# the closest analogue would be an anonymous series like the one shown below:

population = pd.Series([
        2.04, 
        5.23, 
        4.96, 
        2.19, 
        1.29
    ])

area = pd.Series([5386, 
        2058, 
        9993, 
        5905, 
        1826
    
])
print(population)
print(area)
````

````python
# In fact, we can create the same DataFrame with those series, like so:

df = pd.DataFrame({ # the {} indicate a dictionary
                'population': population, # notice this time the values are a series, not a list
                'area': area
               })

df
````

### Accessing Rows

Rows can be accessed by their position using `.iloc` or by their index using `.loc`.

**Using `.iloc`**:

```python
# Access the first row
first_row = df.iloc[0]

# Access the first three rows
first_three_rows = df.iloc[0:3]
```

**Using `.loc`**:

If your DataFrame has a custom index, you can use `.loc` to access rows by their index labels.

```python
# Access the row with an index label of 'a'
row_a = df.loc['a']
```

### Conditional Selection

To filter data based on conditions, you use conditions inside the selection brackets `[]`.

```python
# Select rows where Age is greater than 25
older_than_25 = df[df['Age'] > 25]
```

### Accessing Specific Values

To get a specific value, combine row and column selections using `.loc` or `.iloc`.

```python
# Using .iloc for row and column position
value = df.iloc[2, 1]  # Third row, second column

# Using .loc for index label and column name
value = df.loc[102, 'Name']  # Row with index 102 and column 'Name'
```

### Summary

- **Select columns** by column name.
- **Slice rows** using `.iloc` for positions or `.loc` for index labels.
- **Filter data** with conditions inside brackets `[]`.
- **Access specific values** using `.loc` or `.iloc` with both row and column selectors.
]



#### Adding Columns

You can add a new column by  assigning values to it. If you're adding a column that doesn't exist, it will be created.

```python
df['NewColumn'] = [value1, value2, value3]
```

#### Removing Columns

Use `drop` with the `axis=1` parameter to remove columns (the default is `axis=0`, which will remove rows).

```python
df = df.drop('ColumnName', axis=1)
```

#### Adding Rows

To add a single row, you can use the `append` method with a dictionary that represents the new row.

```python
new_row = {'Column1': value1, 'Column2': value2}
df = df.append(new_row, ignore_index=True)
```


#### Removing Rows

Rows can be removed using the `drop` method, specifying the index of the row(s) you wish to remove.

```python
df = df.drop([index])
```

#### Updating Values

You can update a specific value using the `loc` or `iloc` methods to specify the location.

```python
df.loc[row_index, 'ColumnName'] = new_value
```
