# Significance testing 
## Hypothesis Testing 
 1) Null hypothesis (H₀)
👉 This means “nothing is happening” or “no change”
Example: This teaching method does NOT improve marks

2) Alternative hypothesis (H₁)
👉 This means “something is happening”
 Example: This teaching method improves marks
👉 These two are always opposite ideas


**What we do:** 
* Start with null hypothesis (assume its true)
* Collect some data
* Check the evidence
* Decide:
* If evidence is strong -> Reject H0
* If evedence is weak -> Keep H0 (do not reject)
* We are basically asking: "Does the data give enough proof?"

```python
#import required library 
from scipy import stats 

# Step 1: Sample data (study time in hrs)
data = [2, 3, 2.5, 1.5, 2, 3, 2, 2.2]

# step 2: Define the population mean µ
mu_0 = 2 # Hypothesized mean 

# step 3: Perform one-sample t-test 
t_stat, p_value = stats.ttest_1samp(data, mu_0)

# step 4: Print results 
print("Sample data:", data)
print("Hypothesized mean (H0):", mu_0)
print("T-statistic:", t_stat)
print("P-value:", p_value)

# step 5: Decision making 
alpha = 0.05 # significance level 

if p_value < alpha: 
    print("Decision: Reject the null hypothesis (H0)")
else: 
    print("Decision: Do NOT reject the null hypothesis (H0)")
```
